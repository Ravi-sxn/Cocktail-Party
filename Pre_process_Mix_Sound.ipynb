```python
# To find Sample Rate of .WAV file

import librosa
import librosa.display
import matplotlib.pyplot as plt
import soundfile as sf

def plot_waveform(file_path):
    # Load the audio file
    y, sr = librosa.load(file_path, sr=None)  # Load with original sample rate
    
    # Display sample rate
    print(f"Sample Rate: {sr} Hz")
    
    # Plot waveform
    plt.figure(figsize=(10, 4))
    librosa.display.waveshow(y, sr=sr)
    plt.title("Waveform")
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    plt.show()

# Example usage
file_path_1 = input("Enter Raw Audio File for preprocessing").strip()  # Replace with your file path
plot_waveform(file_path_1)
```

```python
# To convert any sample rate into sample rate 8000 Hz
import librosa
import librosa.display
import matplotlib.pyplot as plt
import soundfile as sf

def plot_waveform(file_path):
    # Load the audio file
    y, sr = librosa.load(file_path, sr=None)  # Load with original sample rate
    
    # Convert sample rate to 8000 Hz
    y_resampled = librosa.resample(y, orig_sr=sr, target_sr=8000)
    sr = 8000  # Update sample rate variable
    
    # Save the resampled audio
    sf.write("/home/planck/Ravi_Saxena/Cocktail_Party/Input_Processed/Resampled.wav", y_resampled, sr)
    
    # Display sample rate
    print(f"Sample Rate: {sr} Hz")
    
    # Plot waveform
    plt.figure(figsize=(10, 4))
    librosa.display.waveshow(y_resampled, sr=sr)
    plt.title("Waveform")
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    plt.show()

# Example usage
file_path =file_path_1  # Replace with your file path
plot_waveform(file_path)
```
